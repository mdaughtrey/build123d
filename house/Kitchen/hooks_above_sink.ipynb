```python
def build_hook():
    hook_width = 30

    base = RectangleRounded(40, hook_width, 5)
    base -= Circle(4).translate((10, 0))
    base = extrude(base, 5)
    surround = extrude(Circle(10/2) - Circle(7/2), amount=3, taper=-1)
    base += surround.translate((10, 0, 5))
    # base += extrude(Circle(10) - Circle(8), amount=5, taper=2).translate((10, 0, 5)
    # base -= CounterSinkHole(radius=4, depth=5, counter_sink_radius=7).translate((10, 0))

    part = RectangleRounded(25, 10, 4.9)
    part -= SlotOverall(15, 3.35).translate((-1, 0))
    part -= Rectangle(12, 11).translate((7, 0))
    part += Rectangle(15, 3.325).translate((8.5, -3.325))
    part = extrude(part, hook_width).rotate(Axis.X, 90)
    # part = fillet(part.edges().filter_by(Axis.Z), radius=1)

    surround = extrude(Circle(10/2) - Circle(7/2), amount=3, taper=-1)
    part += surround.translate((11, -hook_width/2, -1.5))
    part -= Cylinder(radius=2, height=3.325).translate((11, -hook_width/2, -3.325))
    # part = base + Pos(-15, hook_width/2, 10) * part

    return part

```

```python
def build_hook2():
    hook_width = 30
    surround = extrude(Circle(10/2) - Circle(7/2), amount=3, taper=2)

    rounded_top = fillet(Box(8, 3, hook_width).faces().sort_by(Axis.X)[-1].edges(), radius=1)
    rounded_bot = fillet(Box(26, 3, hook_width).faces().sort_by(Axis.X)[-1].edges(), radius=1)
    rounded_bot += surround.rotate(Axis.X, 90).translate((6,3.9, 0))

    part = Cylinder(radius=10/2, height=hook_width) - Cylinder(radius=4/2, height=hook_width) - Box(5, 10, hook_width).translate((3, 0, 0))
    part += rounded_top.translate((4, 3.5, 0))
    part += rounded_bot.translate((13, -3.5, 0))
    part -= Cylinder(radius=2, height=4).rotate(Axis.X, 90).translate((19, -3, 0))

    return part
```

```python
def build_cap():
    part = Cylinder(8, 6)
    cutout = extrude(Circle(5), amount=4, taper=-2)
    part -= cutout.translate((0, 0, -3.2))
    part = fillet(part.edges().sort_by(Axis.Z)[-1], radius=2)
    return part

show(build_cap())
export_step(build_cap(), "/Volumes/SSD 990 2TB/Documents/stl/kitchen_hook_cap.step")
```
